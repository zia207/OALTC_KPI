# COMPREHENSIVE KPI TRACKING & REPORTING FRAMEWORK 


*Prepared by: Zia U. Ahmed, Ph.D. • May 2026*

## EXECUTIVE SUMMARY

In this document, I present a comprehensive, end-to-end framework for defining, tracking, and reporting Key Performance Indicators (KPIs) for OALTC-licensed programs. My framework is structured around five interconnected components: (1) strategic alignment and KPI governance, (2) KPI taxonomy and definition standards, (3) data infrastructure and pipeline architecture using Google BigQuery, (4) dashboard and reporting design, and (5) continuous improvement and stakeholder engagement. Together, these components enable the Bureau of Data Analytics and Automation to deliver timely, accurate, and actionable performance intelligence to OALTC program leaders, senior DOH leadership, and external stakeholders.
 
 I designed this framework to be scalable across all OALTC program types — including nursing homes, assisted living programs, adult day services, home care agencies, and continuing care retirement communities — and aligned it with OALTC's core strategic priorities of quality of care, regulatory compliance, operational efficiency, workforce stability, and health equity.



## **SECTION 1: STRATEGIC ALIGNMENT & KPI GOVERNANCE**

### **1.1 OALTC Strategic Priorities**

All KPIs in this framework are anchored to OALTC's five core strategic priorities. Every metric must trace directly to at least one of these pillars:

|        |                            |                                                                                                               |                                      |
|--------|----------------------------|---------------------------------------------------------------------------------------------------------------|--------------------------------------|
| **\#** | **Strategic Pillar**       | **Description**                                                                                               | **Primary Audience**                 |
| **1**  | **Quality of Care**        | Resident/client outcomes, safety events, care standard adherence across licensed facilities                   | Program Directors, DOH Leadership    |
| **2**  | **Regulatory Compliance**  | Inspection outcomes, deficiency rates, complaint resolution timeliness, enforcement actions                   | Surveyors, Legal, Leadership         |
| **3**  | **Operational Efficiency** | Data submission compliance, processing times, workflow automation, cost-per-case metrics                      | Bureau Chiefs, IT, Finance           |
| **4**  | **Workforce Stability**    | Staff turnover rates, vacancy rates, training completion across regulated providers                           | Program Managers, Workforce Unit     |
| **5**  | **Health Equity**          | Disparity metrics by race, ethnicity, geography, payer type; access and outcome gaps across population groups | Health Equity Team, DOH Commissioner |

### **1.2 KPI Governance Structure**

Robust KPI governance ensures accountability, consistency, and continuous relevance of all metrics. The following three-tier governance model is recommended:

-   Tier 1 — Strategic KPIs: Reviewed by DOH senior leadership and OALTC Director on a quarterly basis. These are high-level outcome indicators tied directly to OALTC's mission and reported in public-facing documents.


-   Tier 2 — Operational KPIs: Reviewed by Bureau Chiefs and Program Managers monthly. These track day-to-day program performance and early warning signals.

-   Tier 3 — Analytical KPIs: Reviewed by the Data Analytics team weekly. These are granular metrics used to monitor data quality, pipeline health, and emerging trends before they escalate.

-   Each KPI is assigned a designated Data Owner (the program office responsible for the metric), a Data Steward (the analytics team member responsible for the data feed), and a Business Reviewer (the stakeholder who interprets and acts on the metric).

-   A KPI Registry — maintained in a version-controlled document linked to the BigQuery metadata catalog — serves as the single source of truth for all active metrics, their definitions, data sources, owners, and update schedules.


## **SECTION 2: KPI TAXONOMY & DEFINITION STANDARDS**

### **2.1 KPI Categories**

KPIs are organized into four functional categories that span the full program performance lifecycle:

-   Outcome KPIs: Measure the end results of care and services (e.g., hospitalization rates, resident satisfaction scores, deficiency citation counts).

-   Process KPIs: Measure whether program activities are being executed as intended (e.g., timely completion of care assessments, inspection cycle adherence).

-   Capacity KPIs: Measure resources and infrastructure (e.g., licensed bed occupancy, staff-to-resident ratios, data system uptime).

-   Equity KPIs: Measure disparity and access gaps across demographic and geographic subgroups (e.g., outcome differentials by race/ethnicity or rural vs. urban location).

### **2.2 Master KPI Definition Template**

Every KPI must be formally defined using the following standardized template before it enters the tracking system. Consistency in definition prevents misinterpretation and ensures comparability over time.

## **SECTION 3: DATA INFRASTRUCTURE & PIPELINE ARCHITECTURE**

### **3.1 Data Architecture Overview**

The KPI tracking system is built on Google BigQuery as the central data warehouse. The architecture follows a four-layer pipeline model: Ingestion → Storage → Transformation → Presentation.

-   Layer 1 — Ingestion: Raw data flows into BigQuery from OALTC-licensed program submissions, ITS-managed applications, Medicaid claims, MDS feeds, and survey system exports. Each data feed is assigned a schema, ingestion schedule, and validation ruleset.

-   Layer 2 — Raw Storage (Data Lake): Unprocessed source data is stored in partitioned BigQuery tables by program, date, and source system. All raw records are retained for auditability and reprocessing.

-   Layer 3 — Transformation (Data Mart): SQL-based ETL pipelines transform raw data into clean, analysis-ready tables. KPI calculations, denominator logic, and exclusion rules are encoded here using version-controlled SQL scripts or dbt models.

-   Layer 4 — Presentation: Transformed KPI tables power dashboards (Looker Studio or Tableau), automated reports, and self-service query interfaces for OALTC program staff.

### **3.2 Data Quality Control Standards**

Data quality is enforced at each pipeline stage using the following six-dimension framework:

|                  |                                                             |                                                                          |                                                                       |
|------------------|-------------------------------------------------------------|--------------------------------------------------------------------------|-----------------------------------------------------------------------|
| **Dimension**    | **Definition**                                              | **Validation Rule (Example)**                                            | **Action if Failed**                                                  |
| **Completeness** | All required fields are populated                           | Null rate \< 2% for facility\_id, report\_date, and outcome fields       | Flag record; alert Data Steward; exclude from KPI calc until resolved |
| **Validity**     | Values fall within defined ranges and formats               | Readmission rate between 0–100%; date format = YYYY-MM-DD                | Reject record; log error; notify submitter                            |
| **Timeliness**   | Data arrives within the expected submission window          | Monthly feeds ingested by the 10th of the following month                | Trigger escalation alert; mark period as provisional                  |
| **Consistency**  | Values are consistent across linked tables and time periods | Facility bed count in submissions matches licensed capacity in registry  | Flag discrepancy; hold for reconciliation                             |
| **Accuracy**     | Values reflect real-world program activity                  | Outlier detection: z-score \> 3 triggers manual review                   | Queue for analyst review; annotate in dashboard                       |
| **Uniqueness**   | No duplicate records in the analysis dataset                | Composite key (facility\_id + resident\_id + event\_date) must be unique | De-duplicate; log count of removed records                            |

### **3.3 KPI Calculation Pipeline (SQL Pattern)**

Each KPI is computed via a standardized SQL view in BigQuery. Below is the generalized template:

**-- KPI: 30-Day Hospital Readmission Rate (Monthly, Facility-Level)**

> CREATE OR REPLACE VIEW oaltc\_kpi.readmission\_rate\_monthly AS
>
> SELECT
>
> facility\_id,
>
> facility\_name,
>
> report\_year,
>
> report\_month,
>
> COUNT(CASE WHEN readmitted\_30d = TRUE THEN 1 END) AS numerator,
>
> COUNT(\*) AS denominator,
>
> ROUND(SAFE\_DIVIDE(COUNT(CASE WHEN readmitted\_30d = TRUE THEN 1 END),
>
> COUNT(\*)) \* 100, 2) AS readmission\_rate\_pct,
>
> CASE
>
> WHEN SAFE\_DIVIDE(...) \* 100 \> 20 THEN 'ALERT'
>
> WHEN SAFE\_DIVIDE(...) \* 100 \> 15 THEN 'WARNING'
>
> ELSE 'ON TARGET'
>
> END AS performance\_status
>
> FROM oaltc\_raw.nh\_discharge\_events
>
> WHERE data\_quality\_flag = 'PASS'
>
> AND discharge\_date \>= DATE\_SUB(CURRENT\_DATE(), INTERVAL 13 MONTH)
>
> GROUP BY 1, 2, 3, 4;


## **SECTION 4: DASHBOARD & REPORTING DESIGN**

### **4.1 Dashboard Hierarchy**

Three tiers of dashboards serve different audiences and decision-making needs:

|                 |                                 |                                   |                                                                              |                   |
|-----------------|---------------------------------|-----------------------------------|------------------------------------------------------------------------------|-------------------|
| **Tier**        | **Dashboard Name**              | **Primary Audience**              | **KPIs Displayed**                                                           | **Refresh Cycle** |
| **Executive**   | **OALTC Strategic Scorecard**   | DOH Commissioner, OALTC Director  | Tier 1 outcome & equity KPIs; trend lines; statewide maps                    | Monthly           |
| **Program**     | **Program Performance Monitor** | Bureau Chiefs, Program Managers   | Tier 2 operational KPIs; facility rankings; alert flags; drill-downs         | Weekly            |
| **Operational** | **Data Operations Dashboard**   | Analytics Team, IT, Data Stewards | Tier 3 data quality KPIs; pipeline status; submission compliance; error logs | Daily             |

**4.2 Standard Reporting Cadence**

The following reporting schedule ensures stakeholders receive the right information at the right time:

-   Daily: Automated data quality alerts sent to Data Stewards for any pipeline failures, validation errors, or submission gaps. No manual action unless thresholds are breached.

-   Weekly: Program Performance Monitor refreshed and shared with Bureau Chiefs. Includes a summary of KPIs in ALERT or WARNING status and any emerging trends.

-   Monthly: Full KPI report package distributed to OALTC Program Directors. Includes trend analysis, month-over-month comparisons, facility-level rankings, and commentary on outliers.

-   Quarterly: Strategic Scorecard presented to DOH senior leadership. Includes progress toward annual targets, equity gap analysis, and narrative summary for each strategic pillar.

-   Annual: Year-end performance review document. Includes baseline resets, target revisions, lessons learned, and proposed additions or retirements to the KPI Registry.

### **4.3 Alert & Escalation Protocol**

A tiered alert system ensures that performance signals translate into timely action:

|                 |                                                                                             |                                                                                  |                                                                                                   |                        |
|-----------------|---------------------------------------------------------------------------------------------|----------------------------------------------------------------------------------|---------------------------------------------------------------------------------------------------|------------------------|
| **Alert Level** | **Trigger Condition**                                                                       | **Notification**                                                                 | **Required Action**                                                                               | **Timeline**           |
| **⚠️ WARNING**  | KPI exceeds 10% above target for one reporting period                                       | Automated email to Program Manager and Data Steward                              | Analyst investigates root cause; documents findings in KPI log                                    | Within 5 business days |
| **🔴 ALERT**     | KPI exceeds warning threshold for two consecutive periods OR single-period breach is severe | Automated alert to Bureau Chief and Program Director; included in weekly summary | Bureau Chief reviews; corrective action plan initiated; escalated to OALTC Director if unresolved | Within 2 business days |
| **✅ ON TARGET** | KPI within or better than target range                                                      | Reflected in dashboard; highlighted in monthly report                            | No action required; document and share as best practice if sustained ≥ 3 months                   | Monthly review         |


## **SECTION 5: ADVANCED ANALYTICS & PREDICTIVE INTELLIGENCE**

### **5.1 Moving Beyond Descriptive KPIs**

While standard KPI tracking captures what has happened, the Bureau of Data Analytics and Automation can deliver significantly greater value by layering predictive and explanatory analytics on top of the core KPI framework. This transforms the system from a rearview mirror into an early-warning and decision-support tool.

**Predictive Modeling**

-   Ensemble machine learning models (Random Forest, XGBoost) can be trained on historical KPI trends to predict which facilities are at elevated risk of deteriorating performance in the next 30–90 days — enabling proactive intervention before a deficiency occurs.

-   Time-series forecasting (ARIMA, Prophet, LSTM) can project KPI trajectories under different intervention scenarios, supporting resource allocation planning.

-   Risk stratification models can flag facilities for prioritized oversight based on a composite risk score combining multiple KPIs, complaint history, and staffing data.

**Explainable AI (XAI) for Root Cause Analysis**

-   SHAP (SHapley Additive exPlanations) values can identify which factors — staffing levels, ownership type, payer mix, geographic context — are the primary drivers of KPI variation across facilities.

-   Geographically Weighted models can reveal whether the relationship between staffing and outcomes differs by region, enabling targeted policy responses rather than one-size-fits-all interventions.

**Health Equity Analytics**

-   Stratified KPI analysis by race, ethnicity, payer type, and geography quantifies disparity gaps and tracks progress toward equity goals over time.

-   Spatial clustering analysis (Moran's I, LISA) can identify geographic hotspots of poor performance, directing inspection and support resources to underserved areas.


## SECTION 6: STAKEHOLDER ENGAGEMENT & CONTINUOUS IMPROVEMENT**

### **6.1 KPI Development Process**

New KPIs are developed through a structured five-step process to ensure relevance, feasibility, and buy-in:

|          |                                 |                                                                                                                                                  |                                   |
|----------|---------------------------------|--------------------------------------------------------------------------------------------------------------------------------------------------|-----------------------------------|
| **Step** | **Activity**                    | **Description**                                                                                                                                  | **Responsible Party**             |
| **1**    | **Use Case Discovery**          | Structured interviews and workshops with program offices to identify analytical needs and performance questions they cannot currently answer     | Data Analyst 3 + Program Managers |
| **2**    | **Data Feasibility Assessment** | Evaluate whether required data exists in BigQuery, the expected data quality, and any gaps that require new data feeds or collection instruments | Data Steward + IT                 |
| **3**    | **KPI Definition & Review**     | Draft formal KPI definition using the standard template; review with program office for accuracy and business relevance                          | Data Analyst 3 + Data Owner       |
| **4**    | **Pilot & Baseline**            | Calculate the KPI for at least 12 months of historical data to establish baseline; review with stakeholders before setting targets               | Data Analyst 3                    |
| **5**    | **Register & Deploy**           | Add approved KPI to the KPI Registry, build the BigQuery view, connect to dashboard, and train program staff on interpretation                   | Data Steward + Data Analyst 3     |

### **6.2 Annual KPI Review Cycle**

The KPI portfolio is reviewed annually to ensure ongoing relevance:

-   Retire KPIs that are no longer aligned with strategic priorities or where the underlying program has changed.

-   Revise targets based on actual performance trends, new benchmarks from CMS or peer states, and updated strategic goals.

-   Add new KPIs identified through the use-case discovery process or driven by new data feeds coming online.

-   Assess data quality trends for each KPI and invest in data improvement efforts for metrics showing persistent quality issues.

### **6.3 Capacity Building & Self-Service Analytics**

A key goal is to build data literacy within OALTC program offices so that staff can independently explore KPI data without relying solely on the analytics team for every question:

-   Develop user guides and short training modules for each dashboard, tailored to the non-technical program officer.

-   Create a curated library of pre-built BigQuery queries that program staff can run directly for common ad-hoc questions.

-   Host quarterly 'Data Office Hours' sessions where program staff can bring analytical questions and learn best practices for interpreting KPI trends.

By combining rigorous KPI governance, a robust BigQuery data architecture, tiered dashboards, and predictive analytics, the Bureau of Data Analytics and Automation can move from reactive reporting to proactive program stewardship — ultimately improving outcomes for New York's aging and long-term care population.
