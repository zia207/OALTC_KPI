# Requirment Gathering


Requirements Gathering is a universal process used across fields and industries to determine and document the specific needs, expectations, and constraints for any new system, project, product, or analysis—before any actual development or implementation begins. The goal is to ensure that, regardless of context, stakeholders’ true problems, goals, and priorities are clearly understood and agreed upon. This structured approach helps teams deliver solutions that are genuinely useful and avoid wasted effort on features or analyses that do not address the real needs of those involved.






## Why it matters

Without requirements gathering, there is a significant risk of building the wrong solution. A classic failure pattern occurs as follows:

> An analyst spends three weeks building a comprehensive dashboard tracking facility inspection completion rates—only to discover that the program manager actually needed **complaint resolution times**, not inspection rates, and already had the inspection data in a spreadsheet.

Requirements gathering prevents issues like this by ensuring alignment between what stakeholders *need*, what data *exists*, and what can *realistically be built*.




## The five core questions to answer

Every requirements gathering effort—whether formal or informal—aims to answer these:

**1. What decision is being made?**
Not "what data is wanted," but what *action* the data will inform. Example: "The program must decide which facilities to prioritize for unannounced inspections next quarter."

**2. What does success look like?**
How will stakeholders know the analysis is working? Example: "If the process can flag the top 10% highest-risk facilities two months before a deficiency citation occurs."

**3. What data is already available?**
What systems, spreadsheets, or reports exist currently? Who owns them? How often are they updated?

**4. What is missing or broken?**
Where are the gaps—missing fields, inconsistent definitions, or data that exists on paper but not digitally?

**5. Who needs to see the output, and in what form?**
For example, a Commissioner may require a one-page summary, a program manager may require a filterable table, and a surveyor may need a facility-level flag. The same underlying data may need to be presented in very different ways depending on the audience.




## How it works in practice — the OALTC context

Requirements gathering in the OALTC context typically proceeds as follows:

**Stage 1 — Stakeholder interviews (Weeks 1–3)**

The project leads or analysts meet with each program office—Nursing Home, Home Care, Adult Day Services, Assisted Living—and conduct structured interviews. They ask questions such as:

- "Walk me through your current reporting process. What do you produce, for whom, and how often?"
- "What question keeps coming up in leadership meetings that you can't currently answer with data?"
- "If you had a dashboard in front of you right now, what would be on it?"
- "What decisions get delayed because you're waiting for data?"
- "What data do you collect that nobody ever uses?"

Every answer is documented, with a focus on listening rather than immediately committing to build solutions.

**Stage 2 — Data inventory (Weeks 2–4)**

At the same time, the team audits existing data resources within BigQuery, ITS-managed applications, and any shadow systems (such as spreadsheets, Access databases, or manual logs). This mapping exercise includes:

- Identifying existing tables, their fields, completeness, and update frequency
- Determining which program needs can be fulfilled with existing data
- Pinpointing needs that would require new data feeds or collection instruments

**Stage 3 — Use case prioritization (Week 4–5)**

Since not all stakeholder requests can be addressed immediately, program leadership collaborates to rank use cases based on: **impact** (importance of the decision being supported) and **feasibility** (ease of building given available data). Those with high impact and high feasibility are prioritized for initial development.

| Use Case                                    | Impact  | Feasibility         | Priority    |
|---                                          |---      |---                 |---          |
| Flag high-risk facilities for inspection     | High    | High               | Build first |
| Predict staffing shortages 90 days out       | High    | Low (data gaps)    | Phase 2     |
| Track complaint resolution by complaint type | Medium  | High               | Build first |
| Benchmark NY facilities vs. national CMS data| Medium  | Medium             | Phase 2     |

**Stage 4 — Requirements documentation**

For each approved use case, a formal requirements document is developed covering:

- **Business requirement**: the stakeholder's need articulated in straightforward language
- **Functional requirement**: the specific actions or computations the system must perform (for example, "calculate 30-day readmission rate at facility level, monthly")
- **Data requirement**: relevant tables, fields, filters, and applicable business rules
- **Output requirement**: the intended audience, expected format, delivery frequency, and depth of detail (drill-down level)
- **Acceptance criteria**: the means by which stakeholders confirm that the deliverable meets their needs

**Stage 5 — Sign-off**

The program office reviews and formally approves the requirements document before any code is written. This process ensures clear agreement on scope and provides a documented record of commitments for both parties.





## Common pitfalls

**Stakeholders articulate wants rather than needs.** Stakeholders may request "all the data in one place," when in fact their actual requirement could be a concise weekly summary of just a few key metrics. It is important to probe deeper by asking, "What would be done differently if that information were available?"

**Requirements creep.** New ideas often emerge during discussions. Without a structured sign-off process, project scope can expand without bounds. Teams should capture emerging ideas in a backlog and evaluate them separately from the current requirements.

**Assuming prior knowledge.** Even with extensive domain expertise, each program office operates with unique workflows, political influences, and definitions. For instance, the definition of "timely" in the nursing home program may differ from that in the home care program. One should never assume; it is important to always ask clarifying questions and document the answers.

**Technical jargon alienating stakeholders.** Use of specialized terminology, such as asking a program manager about "ETL requirements," may stall meaningful conversation. Instead, asking, "How does the data get from the facility to your desk today, and where does it break?" encourages more accessible and productive dialogue.




## Requirements Gathering Interview

**Part 1 — Program Context & Current State** (7 questions) establishes what the program does, to whom it reports, what reports are currently produced, and the quality of the existing data — all prior to requesting any new information.

**Part 2 — Analytical Needs & Decision Support** (8 questions) focuses on core requirements: decisions that cannot currently be made with available data, the contents of a hypothetical dashboard, and equity concerns requiring monitoring.

**Part 3 — Data Specifications & Business Rules** (7 questions) captures precise numerator and denominator definitions, exclusion rules, authoritative data sources, and protocols for handling missing or late submissions — these details prevent KPIs from being built incorrectly.

**Part 4 — Output Design & Reporting Requirements** (7 questions) defines the intended audience, required formats, reporting frequency, and levels of geographic detail for reporting outputs.

**Part 5 — Alert & Escalation Requirements** (4 questions) establishes performance thresholds, notification channels, response timelines, and documentation procedures for resolution.

**Part 6 — Constraints, Risks & Success Criteria** (7 questions) identifies political sensitivities, legal or HIPAA constraints, project timelines, and — critically — the metrics by which project success will be determined by stakeholders.

**Part 7 — KPI Candidate Log** — a structured table to document every KPI identified during the interview, including priority, feasibility, and proposed implementation phase.

**Part 8 — Data Source Inventory** — records each data source along with owner, format, update frequency, BigQuery status, and access requirements.

**Part 9 — Open Items & Next Steps** — logs unresolved issues with assigned owners and due dates, as well as a notes field for the interviewer’s observations.

**Part 10 — Sign-Off** — a formal signature block for the Program Office Representative, Bureau Chief, Data Analyst 3, and Legal/Privacy reviewer to complete before development commences.

## **APPENDIX: INTERVIEWER QUICK REFERENCE**

### **Golden rules during the interview**

-   Interviewers should listen for decisions, not just data requests — they should always ask, "What would you do differently if you had this?"

-   Interviewers should never promise to build anything during the interview — instead, they should capture requests as requirements and assess feasibility separately.

-   If a stakeholder asks for "all the data," interviewers should probe for the underlying question the stakeholder is trying to answer.

-   If a specific chart or report format is named, interviewers should ask what question it answers before agreeing to replicate it.

-   Interviewers should capture verbatim quotes for particularly clear statements of need, as these become powerful justifications in the requirements log.

-   If two stakeholders disagree on a definition or priority, interviewers should document both views and escalate to resolve before building.




### **KPI category quick guide**

|              |                                                         |                                                                      |
|--------------|---------------------------------------------------------|----------------------------------------------------------------------|
| **Category** | **What it measures**                                    | **OALTC example**                                                    |
| Outcome      | Measures the end result of care or program activity     | 30-day readmission rate, pressure ulcer rate, resident satisfaction  |
| Process      | Measures whether activities are executed correctly      | Complaint investigated within 10 days, care plan updated on schedule |
| Capacity     | Measures resources and infrastructure                   | RN hours per resident day, staff turnover, licensed bed occupancy    |
| Equity       | Measures disparity across demographic/geographic groups | Outcome gap by race/ethnicity, rural vs. urban access score          |



### **Common pitfalls to avoid**

|                                 |                                                                                                               |
|---------------------------------|---------------------------------------------------------------------------------------------------------------|
| **Pitfall**                     | **How to handle it**                                                                                          |
| ❌ Stakeholder asks for a metric | Always ask: what decision does this metric inform? If they cannot answer, the metric may not be necessary.    |
| ❌ Scope creep mid-interview     | Capture new ideas in a ‘Parking Lot’ section. Do not agree to include them in the current engagement.         |
| ❌ Data confusion                | Distinguish between what data exists vs. what data is reliable vs. what data is accessible. All three differ. |
| ❌ Vague acceptance criteria     | Every requirement needs a testable definition of done. ‘A better dashboard’ is not a requirement.             |
| ❌ Building before sign-off      | Without sign-off, scope will shift. Protect yourself and the stakeholder by getting written agreement first.  |
